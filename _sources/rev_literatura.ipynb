# **2. Revisión de Literatura**

Este capítulo presenta el estado del arte en segmentación de tumores cerebrales mediante aprendizaje profundo, organizado en tres bloques temáticos: datasets existentes y sus limitaciones, cómo BRISC 2025 supera esas limitaciones y por ultimo, la evolución de las arquitecturas de segmentación desde U-Net hasta los modelos basados en Transformer

---

## 2.1 Datasets Existentes y Sus Limitaciones

El desarrollo de modelos de aprendizaje profundo para segmentación de tumores cerebrales ha dependido históricamente de un conjunto reducido de datasets públicos, cada uno con limitaciones específicas que han condicionado el alcance y la generalización de los modelos propuestos.

### 2.1.1 BraTS

El dataset BraTS {cite}`menze2014`, organizado como un challenge internacional desde 2012, es el recurso más utilizado en la literatura. Ofrece imágenes MRI multimodal (T1, T1ce, T2 y FLAIR) de pacientes con glioma de alto y bajo grado, con anotaciones detalladas de subregiones tumorales. Su escala y estructura competitiva lo han convertido en el estándar de referencia del campo.

Sin embargo, BraTS presenta limitaciones importantes. En primer lugar, **cubre exclusivamente gliomas**, excluyendo otros tipos tumorales prevalentes como meningiomas y tumores pituitarios. En segundo lugar, los datos han sido preprocesados con corrección de inhomogeneidad, registro en espacio estándar y *skull-stripping*, pasos que **no reflejan la variabilidad real** de datos adquiridos en entornos clínicos con diferentes equipos y protocolos. En tercer lugar, las imágenes son **volumétricas 3D**, lo que aumenta sustancialmente los requerimientos computacionales y limita su aplicabilidad en contextos donde solo se dispone de cortes 2D individuales.

### 2.1.2 Dataset de Cheng

El dataset de Cheng {cite}`cheng2017`, disponible en Figshare, contiene imágenes MRI de tres tipos de tumores cerebrales (meningioma, glioma y pituitario) y ha sido ampliamente utilizado como base para trabajos de clasificación multiclase. No obstante, **carece de máscaras de segmentación a nivel de píxel**, lo que lo hace inapropiado para segmentación semántica sin un proceso adicional de anotación. Adicionalmente, presenta desbalance de clases y limitada diversidad en las condiciones de adquisición.

### 2.1.3 Otros datasets públicos

Recursos como SARTAJ {cite}`sartaj2025` y el dataset de Nickparvar {cite}`nickparvar2021` han sido utilizados principalmente para clasificación binaria o multiclase. Aunque accesibles, presentan limitaciones comunes: ausencia de máscaras de segmentación, desbalance de clases e inconsistencias en las etiquetas. La revisión de Musthafa et al. {cite}`musthafa2025` documenta que estas inconsistencias constituyen uno de los principales obstáculos para el desarrollo de modelos generalizables en el campo.

## 2.2 BRISC 2025: Cómo Supera las Limitaciones Anteriores

Fateh et al. {cite}`fateh2026` publicaron BRISC 2025 en enero de 2026 en *Scientific Data* (Nature Portfolio), diseñado explícitamente para atender las limitaciones de los datasets anteriores. Sus características diferenciadoras son las siguientes:

**Cobertura multiclase equilibrada**: incluye cuatro categorías — glioma (1.401 imágenes), meningioma (1.635), pituitario (1.757) y sin tumor (1.207) — con una distribución balanceada que reduce el sesgo hacia clases mayoritarias presente en otros datasets.

**Máscaras de segmentación validadas por expertos**: las máscaras fueron generadas y revisadas iterativamente por un equipo supervisado por una radióloga certificada. El control de calidad reporta un coeficiente Dice medio de 0.924 entre las máscaras iniciales y las verificadas por experto, con solo el 4.8% de imágenes requiriendo corrección {cite}`fateh2026`.

**Representación multiplanar**: el dataset incluye imágenes en los tres planos de adquisición estándar — axial, coronal y sagital — con distribución aproximadamente uniforme (1.550, 1.660 y 1.790 imágenes en entrenamiento respectivamente). Esta característica permite evaluar la generalización de modelos entre planos, un aspecto poco explorado en la literatura.

**Modalidad única T1**: al trabajar con cortes 2D de secuencias T1 con contraste, BRISC 2025 es directamente aplicable a contextos clínicos donde no se dispone de equipos para adquisición volumétrica multimodal, ampliando su relevancia práctica frente a BraTS.

**Split estratificado y curado**: el dataset fue construido desde 7.023 imágenes originales provenientes de Nickparvar {cite}`nickparvar2021`, SARTAJ {cite}`sartaj2025` y otros, mediante eliminación de duplicados, verificación de etiquetas, detección de artefactos y estandarización de dimensiones antes de la división en conjuntos de entrenamiento (5.000) y prueba (1.000) {cite}`fateh2026`.

## 2.3 Evolución de las Arquitecturas de Segmentación

### 2.3.1 U-Net: el punto de partida

La arquitectura U-Net, propuesta por Ronneberger, Fischer y Brox {cite}`ronneberger2015` en MICCAI 2015, estableció el paradigma dominante en segmentación de imágenes biomédicas. Su estructura encoder-decoder simétrica con *skip connections* entre niveles correspondientes permite combinar información semántica de alta abstracción con información espacial de alta resolución, logrando segmentaciones precisas incluso con conjuntos de entrenamiento pequeños mediante técnicas de aumento de datos.

En el benchmark de BRISC 2025, U-Net alcanzó un mIoU ponderado de 75.7% {cite}`fateh2026`, siendo el modelo con mejor relación rendimiento-complejidad entre las arquitecturas clásicas evaluadas. Este resultado, obtenido con una arquitectura de 2015, ilustra tanto la solidez del diseño original como el margen de mejora disponible con técnicas más recientes.

### 2.3.2 Transformers en segmentación médica

La irrupción de los modelos Transformer en visión por computador impulsó su adopción en segmentación médica. A diferencia de las CNN, cuyo campo receptivo crece de forma local capa a capa, los Transformers modelan dependencias de largo alcance en la imagen mediante mecanismos de auto-atención global, una capacidad especialmente valiosa para tumores con formas irregulares o de gran extensión.

Cao et al. {cite}`cao2022` propusieron **Swin-Unet**, un encoder-decoder puramente basado en Swin Transformer con ventanas deslizantes desplazadas, que demostró superar a las arquitecturas CNN puras en tareas de segmentación de órganos. En el benchmark de BRISC 2025, el mejor resultado fue alcanzado por SaberNet, una arquitectura multi-escala basada en Transformer, con un mIoU ponderado de 80.6% {cite}`fateh2026`, lo que establece el techo de referencia que este trabajo busca superar.

### 2.3.3 Trabajos recientes sobre tumores cerebrales

**Gnana Jenifer y Rajakumar** {cite}`jenifer2026` presentaron **HyBiUnet-TransMask**, un sistema integrado de segmentación y clasificación que combina un encoder híbrido CNN-Swin Transformer con un módulo de desacoplamiento (*Disentanglement Learning Module*) y un autoencoder enmascarado basado en recuperación de casos (RP-MaskAE). Evaluado en BraTS 2023, alcanza un coeficiente Dice de hasta 95.80% en la región tumoral total. Aunque opera en modalidad 3D multimodal, su estrategia de fusión de características entre CNN y Transformer es directamente relevante como referencia arquitectónica para este trabajo.

**Zhu et al.** {cite}`zhu2024` desarrollaron **SDV-TUNet**, una red Transformer 3D con atención dispersa dinámica (*Sparse Dynamic Attention*) y un módulo de fusión de características de bordes en múltiples niveles (MEFF). Sus resultados en BraTS 2020 (Dice 87.30%) y BraTS 2021 (Dice 90.58%) demuestran la efectividad de los módulos de fusión de bordes para capturar regiones peritumorales con precisión. El módulo MEFF constituye una inspiración concreta para el componente novedoso del modelo propuesto en el Capítulo 7.

**Mehmood et al.** {cite}`mehmood2025` abordaron un problema clínico crítico: la degradación del rendimiento cuando ciertas secuencias MRI no están disponibles. Propusieron un marco basado en Tri-Factorización de Matrices No Negativas (NMTF) para reconstruir modalidades faltantes, demostrando una mejora promedio de ≈10% en exactitud sobre U-Net, V-Net y DenseNet. Aunque la estrategia no aplica directamente a este trabajo (que opera con una sola modalidad), el análisis de robustez ante datos incompletos es relevante para la discusión de limitaciones.

**Ali et al.** {cite}`ali2025` propusieron un sistema para segmentación y clasificación de seis tipos de tumores cerebrales mediante segmentación ER-BHS combinada con características híbridas y clasificadores de aprendizaje automático, alcanzando una exactitud del 98.61%. Aunque emplea una metodología diferente a la segmentación semántica profunda, constituye un antecedente relevante sobre el manejo de múltiples clases tumorales en MRI 2D.

La revisión comprehensiva de **Musthafa et al.** {cite}`musthafa2025` sintetiza más de dos décadas de investigación e identifica como desafíos persistentes el desbalance de clases, la heterogeneidad de datos inter-institucional y la falta de validación clínica externa, desafíos que BRISC 2025 aborda parcialmente con su diseño balanceado y multiplanar.
